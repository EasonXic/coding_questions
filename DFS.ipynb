## 1. 组合数

```python
"""
DFS (Depth-First Search) 深度优先搜索

dfs(path, x, k)
path: current search path
x: current node or position
k: additional parameter for search constraints e.g. depth

"""
```

### [1,n]其中k个数的组合(不重复 无序) = C(k,n)

```python
def combinations(nums, K):
    if len(nums) < K or K==0:
        return 0
    
    res_paths = [] # record all the possible search ppath
    def dfs(path, x, k):
        # path: current search path
        # x: current node or position
        # k: left depth
        if k == 0:
            res_paths.append(path)
            return
        for i in range(x+1, len(nums)):
            dfs(path + [nums[i]], i, k - 1)

    dfs([], -1, K)
    print(res_paths)
    return len(res_paths)


nums = range(5)
K=3
combinations(nums, K)
```

```python
# nums 长度为n，sum为target的组合数(不超过k个数)
# assume nums is a list of positive integers and target is a positive integer

def combinations(nums, K, target):
    if len(nums) < K or K==0:
        return 0
    
    res_paths = [] # record all the possible search ppath
    def dfs(path, x, k, m):
        # path: current search path
        # x: current node or position
        # k: left depth
        # m: remaining target sum
        if m==0:
            if k>=0:
                res_paths.append(path) # record the valid combination
            return
        if m < 0:
            return
        
        for i in range(x+1, len(nums)):
            dfs(path + [nums[i]], i, k - 1, m - nums[i])

    dfs([], -1, K, target)
    print(res_paths)
    return len(res_paths)


nums = range(1,6)
K=3
target = 6
combinations(nums, K, target)
```

### [1,n]其中k个数的组合(不重复 有序) = A(k,n)


```python
def combinations(nums, k):
    if len(nums) < k or k==0:
        return 0
    
    res_paths = [] # record all the possible search ppath
    def dfs(path, x, k):
        # path: current search path
        # x: current node or position
        # k: left depth
        if k == 0:
            res_paths.append(path)
            return
        for i in range(0, len(nums)):
            if nums[i] not in path:
                dfs(path + [nums[i]], i, k - 1)

    dfs([], -1, k)
    print(res_paths)
    return len(res_paths)


nums = range(3)
K=3
combinations(nums, K)
```

### [1,n]其中k个数的组合(可重复 无序)

```python
def combinations(nums, k):
    if len(nums) < k or k==0:
        return 0
    
    res_paths = [] # record all the possible search ppath
    def dfs(path, x, k):
        # path: current search path
        # x: current node or position
        # k: left depth
        if k == 0:
            res_paths.append(path)
            return
        
        for i in range(x, len(nums)):
            dfs(path + [nums[i]], i, k - 1)

    # start DFS from each element in nums
    for i in range(len(nums)):
        dfs([nums[i]], i, k - 1)

    print(res_paths)
    return len(res_paths)


nums = range(3)
K=3
combinations(nums, K)
```

## 2. N皇后

```python
# N皇后  O(n!)因为回溯其实要遍历一个全排列
def isValid(pos, row, col):  # pos下标为行号，元素为列号，记录皇后位置，显然现在只有前row行摆放了棋子  row是已经遍历到第几行
    # 遍历所有已经记录的行
    for i in range(row): # 显然现在只有前row行摆放了棋子，因此检测前row行的所有棋子，是否与位置row, col冲突
        # 不能同行同列同一斜线
        if row == i or col == pos[i] or abs(row - i) == abs(col - pos[i]):
            return False
    return True
def rec(n, row, pos, res): # 递归查找皇后种类  n棋盘大小  row已经遍历到第几行  pos下标为行号，元素为列号，记录皇后位置   res计数
    if row == n:  # 完成全部行都选择了位置 返回
        res += 1
        return int(res)

    for i in range(n):  # 遍历所有列，当发现此时在第row行和第i列可以摆放棋子，就给该位置摆放一个棋子，并从下一行继续递归
        # 检查该位置是否符合条件
        if isValid(pos, row, i): # 检查现在摆放的所有棋子，是否跟第row行和第i列冲突
            # 加入位置
            pos[row] = i
            # 递归继续查找
            res = rec(n, row + 1, pos, res)
    return res
def Nqueen(n): # n棋盘大小
    res = 0
    pos = [0] * n # 下标为行号，元素为列号，记录皇后位置，一开始全是0表示不放棋子
    # 递归
    result = rec(n, 0, pos, res) # 从第0行开始，也就是从没有棋子开始
    return result

print(Nqueen(8))
```
