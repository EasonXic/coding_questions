## 1. 滑动窗口最大值：list/circle

```python
# 给你一个整数数组 nums，有一个大小为 k 的滑动窗口从数组的最左侧移动到数组的最右侧。你只可以看到在滑动窗口内的 k 个数字。滑动窗口每次只向右移动一位。
# 返回 滑动窗口中的最大值：一个list


"""
用单调队列解

- 单调队列里面的元素是递减的，从左到右，所以队列的左端永远是当前窗口的最大值
- 从左到右元素的age依次减小（越靠右的越新）

每当窗口移动时：
1. 会放进来一个新的element，如果这个element比队列中现有的某个元素i大，那么i以及i右边deque中的元素都会被移除，因为它们不可能再成为窗口的最大值（比这个新element小，而且比它老）
2. 检查队列里最老的元素（最左边的）是否已经过期（不在当前窗口内），如果过期就从左边pop掉

"""

# 滑动窗口最大值 O(n)  space O(size)
from collections import deque

def maxSlidingWindow(num, size):
    res = []
    if not num or size == 0:
        return res
    if size > len(num):
        return [max(num)]
    
    dq = deque()  # dq的目的是存一个当前窗口下，最大元素对应位置，当我们添加元素的时候，先比较队列中的元素大小和新元素的大小，假如比新元素小一律pop出
    # 因为这意味着现在有一个更大的元素进入当前的窗口，这些元素都作废了
    # 同时还要记得去维护一下队列里的元素有没过期，也就是被窗口经过了
    # 这个dq只记录元素的位置

    # 初始化第一个队列
    for i in range(size): 
        # 生成第一个dq 如果新加入的element比已有的元素大，就把这些元素都pop掉
        while len(dq) != 0 and num[dq[-1]] < num[i]:
            dq.pop()
        dq.append(i)

    # main loop for the rest of the array
    for i in range(size, len(num)):  # i表示当前window的最后一个元素位置 每次都维护一个dq,dq[0]对应当前window的最大元素位置
        res.append(num[dq[0]]) # record the maximum of the current window
        
        while len(dq) != 0 and dq[0] < (i - size + 1):  # 在每轮迭代中，检查dq[0]是否还在当前window里面，如果已经不在了就从左边pop
            dq.popleft() # 或者假如队列已经是空的了，那说明原来的最后一个元素是最大的，现在不在窗口里了
        
        while len(dq) != 0 and num[dq[-1]] < num[i]:  # 在每轮迭代中，在加入新的元素之前，把队列里面比它小的剔除，这样得到的dq永远是递减的（从左到右）
            dq.pop() # 或者假如队列已经是空的了，那说明这个新的元素比窗口里的其他元素都大
        
        dq.append(i) # 新元素入队

    res.append(num[dq[0]]) # 最后一个窗口
    return res


num=[2,3,4,2,6,2,5,1]
size=3
print(maxSlidingWindow(num, size))


```

```python
# 环形

"""
只需要将数组首尾相连，形成一个新的数组，然后在这个新的数组上使用滑动窗口最大值的算法即可。
比如
num=[2,3,4,2,6,2,5,1]
size=3

new_num = num + num[:size-1]  # 将数组首尾相连，形成一个新的数组
"""

from collections import deque
def maxSlidingWindow(num, size):
    res = []
    if not num or size == 0:
        return res
    if size > len(num):
        return [max(num)]

    num_enriched = num + num[:size-1]  # 将数组首尾相连，形成一个新的数组
    dq = deque()
    for i in range(size): 
        while len(dq) != 0 and num_enriched[dq[-1]] < num_enriched[i]:
            dq.pop()
        dq.append(i)

    for i in range(size, len(num_enriched)):
        res.append(num_enriched[dq[0]])
        while len(dq) != 0 and dq[0] < (i - size + 1):
            dq.popleft()
        while len(dq) != 0 and num_enriched[dq[-1]] < num_enriched[i]:
            dq.pop()
        dq.append(i)

    res.append(num_enriched[dq[0]])
    return res


num=[2,3,4,2,6,2,5,1]
size=3
print(maxSlidingWindow(num, size))
```

## 3. 归并

```python
# 数组中的逆序对  O(nlogn)就是归并排序的复杂度
mod = 1000000007
# 本质上可以用归并的思想，就是把原来的数组划分成左右两个子串，然后数组整体的逆序对就等于左边的和右边的子串逆序对之和，再加上左边比右边大的数的个数
# 在统计左边比右边大的数的个数的时候其实还是很麻烦，得遍历，但如果子串都是有序的就很方便，就仅是线性复杂度，因此直接调用归并排序的思路，归并的同时统计逆序对
def MergeSort(left: int, right: int, data, temp) -> int:
    # 停止划分
    if left >= right:
        return 0
    # 取中间
    mid = int((left + right) / 2)
    # 左右划分合并
    res = MergeSort(left, mid, data, temp) + MergeSort(mid + 1, right, data, temp)
    # 防止溢出
    res %= mod
    i, j = left, mid + 1
    for k in range(left, right + 1):
        temp[k] = data[k]
    # 这里直接用归并排序的merge操作，把原数组排序，与此同时统计逆序对
    for k in range(left, right + 1):
        if i == mid + 1:
            data[k] = temp[j] # 归并
            j += 1
        elif j == right + 1 or temp[i] <= temp[j]:
            data[k] = temp[i]
            i += 1
        else: # 左边比右边大，答案增加
            data[k] = temp[j]
            j += 1
            res += mid - i + 1 # 统计逆序对
    return res % mod

def InversePairs(data) -> int:
    n = len(data)
    res = [0]*n # 一个辅助数组
    return MergeSort(0, n - 1, data, res)

print(InversePairs([1,2,3,4,5,6,7,0]))
```

```python
# 滑动窗口中位数 O(nlogk)
# 这个题我记得没有什么特殊的技巧，主要用的优化方法还是需要用一个二分查找的库，来把维护有序窗口的复杂度降为O(logk)
def medianSlidingWindow(nums, k: int):
    import bisect # 使用二分查找库
    median = lambda a: (a[(len(a) - 1) // 2] + a[len(a) // 2]) / 2  # 用于对一个有序数组，计算其中位数
    a = sorted(nums[:k]) # a记录当前数组，最开始等于排好序的 nums 的前 k 个元素
    res = [median(a)]
    for i, j in zip(nums[:-k], nums[k:]): # i[0:n-k] j[k:n]
        a.pop(bisect.bisect_left(a, i)) # 将值为 i 的索引从 a 中删除  O(logk)
        bisect.insort_left(a, j) # 将j增加在合适的位置 O(logk)
        res.append(median(a))
    return res
```
